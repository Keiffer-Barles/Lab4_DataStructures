```python
LAST_NAME = "Barles"
SEED_NUM = 8
FAVORITE_ARTIST = "FRANK SINATRA"

test_sources = [LAST_NAME + "_T1", LAST_NAME + "_T2", LAST_NAME + "_T3", LAST_NAME + "_T4", LAST_NAME + "_T5"]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

def generate_value(seed, artist, category_index):
    char_sum = 0
    for c in artist:
        char_sum = char_sum + ord(c)
    base = (seed * category_index) + (char_sum % 100)
    return round(base + (category_index * 3.7), 2)

records = []
for source in test_sources:
    k = 1
    for category in categories:
        val = generate_value(SEED_NUM, FAVORITE_ARTIST, k)
        records.append([source, category, val])
        k = k + 1

records.append([LAST_NAME + "_T1", "TEMPERATURE", generate_value(SEED_NUM, FAVORITE_ARTIST, 1)])

extra_cats = ["TEMPERATURE", "VOLTAGE"]
extra_sources = [LAST_NAME + "_T4", LAST_NAME + "_T5"]
for src in extra_sources:
    m = 1
    for cat in extra_cats:
        val = round(SEED_NUM * (m + 2.5), 2)
        records.append([src, cat, val])
        m = m + 1

organized = {}
for rec in records:
    src = rec[0]
    cat = rec[1]
    val = rec[2]
    if src not in organized:
        organized[src] = {}
    if cat not in organized[src]:
        organized[src][cat] = []
    organized[src][cat].append(val)

seen = []
repeated = []
distinct = []
for rec in records:
    key = rec[0] + "|" + rec[1]
    if key in seen:
        repeated.append(rec)
    else:
        distinct.append(rec)
        seen.append(key)

numeric_summary = {}
for cat in categories:
    vals = []
    for rec in records:
        if rec[1] == cat:
            vals.append(rec[2])
    count = len(vals)
    min_val = min(vals)
    max_val = max(vals)
    avg_val = sum(vals) / count
    numeric_summary[cat] = [count, round(min_val, 2), round(max_val, 2), round(avg_val, 2)]

ordered_records = sorted(records)

print("=" * 60)
print("ENGINEERING TEST DATA ORGANIZER — FINAL SUMMARY")
print("Student:", LAST_NAME, "| Seed:", SEED_NUM, "| Artist:", FAVORITE_ARTIST)
print("=" * 60)

print("\n[1] GENERATED TEST DATA — SOURCES")
print("Test Sources:", test_sources)
print("Categories:", categories)

print("\n[2] ORGANIZED TEST DATA")
for src in organized:
    print(src)
    for cat in organized[src]:
        print("  ", cat, ":", organized[src][cat])

print("\n[3] REPEATED AND DISTINCT DATA")
print("Distinct Entries:", len(distinct))
print("Repeated Entries:", len(repeated))
if len(repeated) > 0:
    print("Repeated Record(s):")
    for r in repeated:
        print("  Source:", r[0], "| Category:", r[1], "| Value:", r[2])

print("\n[4] NUMERICAL AND CATEGORICAL SUMMARIES")
print("--- Numerical Summary by Category ---")
print("CATEGORY".ljust(15), "COUNT".rjust(6), "MIN".rjust(8), "MAX".rjust(8), "AVERAGE".rjust(8))
for cat in numeric_summary:
    stats = numeric_summary[cat]
    print(cat.ljust(15), str(stats[0]).rjust(6), str(stats[1]).rjust(8), str(stats[2]).rjust(8), str(stats[3]).rjust(8))

print("\n--- Categorical Summary ---")
print("Total Records Collected:", len(records))
print("Unique Source-Category Pairs:", len(seen))
print("Number of Categories:", len(categories))
print("Number of Test Sources:", len(test_sources))
print("Duplicate Entries Found:", len(repeated))

print("\n[5] ORDERED RESULTS")
print("SOURCE".ljust(14), "CATEGORY".ljust(16), "MEASURED VALUE")
print("-" * 50)
for rec in ordered_records:
    print(rec[0].ljust(14), rec[1].ljust(16), rec[2])

print("\n[6] PROCESSED AND FORMATTED RESULTS")
print("Total Measurements:", len(records))
print("Distinct Measurements:", len(distinct))
print("Repeated Measurements:", len(repeated))
print("Data organized by source then category — sorted alphabetically.")

print("\n[7] FINAL CONSOLIDATED SUMMARY")
print("Student Name:", LAST_NAME)
print("Seed Value:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
print("Categories Analyzed:", ", ".join(categories))
print("Test Sources Used:", ", ".join(test_sources))
print("Total Data Points:", len(records))
print("Unique Combinations:", len(seen))
print("Duplicate Entries:", len(repeated))
print("Status: Complete — all data generated, organized, and summarized.")
print("=" * 60)
```

```python
LAST_NAME = "Barles"
SEED_NUM = 8
FAVORITE_ARTIST = "FRANK SINATRA"

student_labels = ["S1", "S2", "S3", "S4", "S5"]
student_names = []
for label in student_labels:
    student_names.append(LAST_NAME + "_" + label)

subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

def generate_score(seed, artist, pos):
    art_len = len(artist)
    index = pos % art_len
    char_val = ord(artist[index])
    score = (seed * 10) + (char_val % 40)
    if score > 100:
        score = 95
    return round(score, 1)

def get_category(score):
    if score is None:
        return "INCOMPLETE"
    if score >= 90:
        return "OUTSTANDING"
    elif score >= 80:
        return "VERY SATISFACTORY"
    elif score >= 70:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

records = []
pos = 0
for name in student_names:
    for subj in subjects:
        score = generate_score(SEED_NUM, FAVORITE_ARTIST, pos)
        records.append([name, subj, score])
        pos = pos + 1

records.append([LAST_NAME + "_S1", "PROGRAMMING", generate_score(SEED_NUM, FAVORITE_ARTIST, 0)])

records.append([LAST_NAME + "_S3", "ENGLISH", None])

organized = {}
for rec in records:
    name = rec[0]
    subj = rec[1]
    scr = rec[2]
    if name not in organized:
        organized[name] = {}
    organized[name][subj] = scr

seen_pairs = []
repeated_records = []
incomplete_records = []
distinct_records = []
for rec in records:
    key = rec[0] + "|" + rec[1]
    if rec[2] is None:
        incomplete_records.append(rec)
    if key in seen_pairs:
        repeated_records.append(rec)
    else:
        seen_pairs.append(key)
        distinct_records.append(rec)

categorized = []
for rec in records:
    cat = get_category(rec[2])
    categorized.append([rec[0], rec[1], rec[2], cat])

student_summary = {}
for name in student_names:
    scores = []
    for rec in records:
        if rec[0] == name and rec[2] is not None:
            scores.append(rec[2])
    if len(scores) > 0:
        avg_val = sum(scores) / len(scores)
        student_summary[name] = [len(scores), round(avg_val, 2), max(scores), min(scores)]
    else:
        student_summary[name] = [0, 0, 0, 0]

subject_summary = {}
for subj in subjects:
    scores = []
    for rec in records:
        if rec[1] == subj and rec[2] is not None:
            scores.append(rec[2])
    if len(scores) > 0:
        avg_val = sum(scores) / len(scores)
        subject_summary[subj] = [len(scores), round(avg_val, 2), max(scores), min(scores)]
    else:
        subject_summary[subj] = [0, 0, 0, 0]

def safe_sort_key(record):
    score = record[2]
    if score is None:
        score = -1
    return (record[0], record[1], score)

ordered_results = sorted(categorized, key=safe_sort_key)

print("=" * 65)
print("STUDENT PERFORMANCE DATA ANALYZER — FINAL REPORT")
print("Student:", LAST_NAME, "| Seed:", SEED_NUM, "| Artist:", FAVORITE_ARTIST)
print("=" * 65)

print("\n[1] GENERATED PERFORMANCE RECORDS")
print("STUDENT".ljust(16), "SUBJECT".ljust(14), "SCORE")
print("-" * 45)
for rec in records:
    if rec[2] is None:
        score_str = "MISSING"
    else:
        score_str = str(rec[2])
    print(rec[0].ljust(16), rec[1].ljust(14), score_str)

print("\n[2] ORGANIZED STUDENT/SUBJECT DATA")
for name in organized:
    print("\n" + name)
    for subj in organized[name]:
        val = organized[name][subj]
        if val is None:
            val_str = "MISSING"
        else:
            val_str = str(val)
        print("  " + subj.ljust(14), val_str)

print("\n[3] REPEATED AND INCOMPLETE RECORDS")
print("Repeated Entries:", len(repeated_records))
if len(repeated_records) > 0:
    for r in repeated_records:
        print("  REPEATED:", r[0], "-", r[1], "Score:", r[2])
print("Incomplete Entries:", len(incomplete_records))
if len(incomplete_records) > 0:
    for r in incomplete_records:
        print("  INCOMPLETE:", r[0], "-", r[1], "Score: MISSING")

print("\n[4] PERFORMANCE CATEGORIES")
print("STUDENT".ljust(16), "SUBJECT".ljust(14), "SCORE".ljust(8), "CATEGORY")
print("-" * 65)
for rec in categorized:
    if rec[2] is None:
        score_str = "N/A"
    else:
        score_str = str(rec[2])
    print(rec[0].ljust(16), rec[1].ljust(14), score_str.ljust(8), rec[3])

print("\n[5] STUDENT AND SUBJECT SUMMARIES")
print("--- Student Summary ---")
print("STUDENT".ljust(16), "COURSES".rjust(7), "AVG".rjust(8), "HIGH".rjust(8), "LOW".rjust(8))
for name in student_summary:
    s = student_summary[name]
    print(name.ljust(16), str(s[0]).rjust(7), str(s[1]).rjust(8), str(s[2]).rjust(8), str(s[3]).rjust(8))

print("\n--- Subject Summary ---")
print("SUBJECT".ljust(16), "STUDENTS".rjust(7), "AVG".rjust(8), "HIGH".rjust(8), "LOW".rjust(8))
for subj in subject_summary:
    s = subject_summary[subj]
    print(subj.ljust(16), str(s[0]).rjust(7), str(s[1]).rjust(8), str(s[2]).rjust(8), str(s[3]).rjust(8))

print("\n[6] ORDERED RESULTS")
print("STUDENT".ljust(16), "SUBJECT".ljust(14), "SCORE".ljust(8), "CATEGORY")
print("-" * 65)
for rec in ordered_results:
    if rec[2] is None:
        score_str = "N/A"
    else:
        score_str = str(rec[2])
    print(rec[0].ljust(16), rec[1].ljust(14), score_str.ljust(8), rec[3])

print("\n[7] FINAL STUDENT PERFORMANCE REPORT")
print("Student Name:", LAST_NAME)
print("Total Records Generated:", len(records))
print("Unique Student-Subject Pairs:", len(seen_pairs))
print("Repeated Records:", len(repeated_records))
print("Incomplete Records:", len(incomplete_records))
print("Subjects Analyzed:", ", ".join(subjects))
print("Students Evaluated:", ", ".join(student_names))
print("Status: Report Complete — All data generated, categorized, and summarized.")
print("=" * 65)
```

```python
LAST_NAME = "Barles"
SEED_NUM = 8
FAVORITE_ARTIST = "FRANK SINATRA"

customer_labels = ["C1", "C2", "C3", "C4", "C5"]
customers = []
for label in customer_labels:
    customers.append(LAST_NAME + "_" + label)

products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]

prices = {}
prices["NOTEBOOK"] = 45.00
prices["PEN"] = 12.50
prices["USB"] = 35.00
prices["HEADPHONES"] = 180.00
prices["CALCULATOR"] = 230.00

def generate_quantity(seed, artist, pos):
    art_len = len(artist)
    index = pos % art_len
    char_val = ord(artist[index])
    qty = (seed + (char_val % 5)) + 1
    return qty

records = []
pos = 0
for cust in customers:
    for prod in products:
        qty = generate_quantity(SEED_NUM, FAVORITE_ARTIST, pos)
        price = prices[prod]
        value = round(qty * price, 2)
        records.append([cust, prod, qty, price, value])
        pos = pos + 1

first_cust = customers[0]
first_prod = products[0]
first_qty = generate_quantity(SEED_NUM, FAVORITE_ARTIST, 0)
first_price = prices[first_prod]
first_value = round(first_qty * first_price, 2)
records.append([first_cust, first_prod, first_qty, first_price, first_value])

organized = {}
for rec in records:
    cust = rec[0]
    prod = rec[1]
    if cust not in organized:
        organized[cust] = {}
    if prod not in organized[cust]:
        organized[cust][prod] = []
    organized[cust][prod].append([rec[2], rec[3], rec[4]])

seen_pairs = []
repeated_purchases = []
distinct_pairs = []
for rec in records:
    key = rec[0] + "|" + rec[1]
    if key in seen_pairs:
        repeated_purchases.append(rec)
    else:
        seen_pairs.append(key)
        distinct_pairs.append(rec)

customer_summary = {}
for cust in customers:
    total_qty = 0
    total_value = 0
    prod_count = 0
    for rec in records:
        if rec[0] == cust:
            total_qty = total_qty + rec[2]
            total_value = total_value + rec[4]
    unique_prods = []
    for rec in distinct_pairs:
        if rec[0] == cust and rec[1] not in unique_prods:
            unique_prods.append(rec[1])
    prod_count = len(unique_prods)
    customer_summary[cust] = [prod_count, total_qty, round(total_value, 2)]

product_summary = {}
for prod in products:
    total_qty = 0
    total_value = 0
    buyer_count = 0
    for rec in records:
        if rec[1] == prod:
            total_qty = total_qty + rec[2]
            total_value = total_value + rec[4]
    unique_buyers = []
    for rec in distinct_pairs:
        if rec[1] == prod and rec[0] not in unique_buyers:
            unique_buyers.append(rec[0])
    buyer_count = len(unique_buyers)
    product_summary[prod] = [buyer_count, total_qty, round(total_value, 2)]

def sort_key(record):
    return (record[0], record[1])

ordered_results = sorted(records, key=sort_key)

grand_total_qty = 0
grand_total_value = 0
for rec in records:
    grand_total_qty = grand_total_qty + rec[2]
    grand_total_value = grand_total_value + rec[4]

print("=" * 70)
print("TRANSACTION HISTORY ANALYZER — CONSOLIDATED REPORT")
print("Student:", LAST_NAME, "| Seed:", SEED_NUM, "| Artist:", FAVORITE_ARTIST)
print("=" * 70)

print("\n[1] GENERATED TRANSACTION RECORDS")
print("CUSTOMER".ljust(14), "PRODUCT".ljust(14), "QTY".rjust(5), "PRICE".rjust(9), "VALUE".rjust(10))
print("-" * 70)
for rec in records:
    print(rec[0].ljust(14), rec[1].ljust(14), str(rec[2]).rjust(5), str(rec[3]).rjust(9), str(rec[4]).rjust(10))

print("\n[2] ORGANIZED TRANSACTION DATA")
for cust in organized:
    print("\n" + cust)
    for prod in organized[cust]:
        print("  " + prod)
        for entry in organized[cust][prod]:
            print("    Qty:", str(entry[0]).ljust(3), "Price:", str(entry[1]).ljust(7), "Value:", str(entry[2]))

print("\n[3] REPEATED AND DISTINCT TRANSACTIONS")
print("Total Records:", len(records))
print("Distinct Customer-Product Pairs:", len(distinct_pairs))
print("Repeated Purchases:", len(repeated_purchases))
if len(repeated_purchases) > 0:
    for r in repeated_purchases:
        print("  REPEATED:", r[0], "→", r[1], "Qty:", r[2], "Value:", r[4])

print("\n[4] TRANSACTION VALUES SUMMARY")
print("Total Quantity Purchased:", grand_total_qty)
print("Overall Transaction Value:", round(grand_total_value, 2))

print("\n[5] CUSTOMER SUMMARIES")
print("CUSTOMER".ljust(16), "PRODUCTS".rjust(8), "ITEMS".rjust(8), "TOTAL VALUE".rjust(12))
for cust in customer_summary:
    s = customer_summary[cust]
    print(cust.ljust(16), str(s[0]).rjust(8), str(s[1]).rjust(8), str(s[2]).rjust(12))

print("\n[6] PRODUCT SUMMARIES")
print("PRODUCT".ljust(16), "BUYERS".rjust(8), "SOLD".rjust(8), "TOTAL SALES".rjust(12))
for prod in product_summary:
    s = product_summary[prod]
    print(prod.ljust(16), str(s[0]).rjust(8), str(s[1]).rjust(8), str(s[2]).rjust(12))

print("\n[7] ORDERED RESULTS")
print("CUSTOMER".ljust(14), "PRODUCT".ljust(14), "QTY".rjust(5), "PRICE".rjust(9), "VALUE".rjust(10))
print("-" * 70)
for rec in ordered_results:
    print(rec[0].ljust(14), rec[1].ljust(14), str(rec[2]).rjust(5), str(rec[3]).rjust(9), str(rec[4]).rjust(10))

print("\n[8] OVERALL TRANSACTION SUMMARY")
print("Student Name:", LAST_NAME)
print("Total Transactions:", len(records))
print("Unique Customer-Product Combinations:", len(distinct_pairs))
print("Repeated Purchases Found:", len(repeated_purchases))
print("Total Items Sold:", grand_total_qty)
print("Grand Total Revenue:", round(grand_total_value, 2))
print("Products Tracked:", ", ".join(products))
print("Customers Included:", ", ".join(customers))
print("Status: Report Complete — All data generated, analyzed, and summarized.")
print("=" * 70)
```
